```python
import numpy as np
import matplotlib.pyplot as plt
import os

plt.rcParams.update({
    'font.size': 16,
    'text.usetex': True,
    'axes.titlesize': 'large',  # Title size
    'axes.titleweight': 'bold',  
    'axes.labelsize': 'LARGE',  
    'axes.labelweight': 'bold',  
    'xtick.labelsize': 'medium',  
    'ytick.labelsize': 'medium',  
    'text.latex.preamble': r'\boldmath',
})

plt.rcParams['xtick.direction'] = 'in'
plt.rcParams['ytick.direction'] = 'in'
plt.rcParams['xtick.major.size'] = 10
plt.rcParams['ytick.major.size'] = 10
plt.rcParams['xtick.minor.size'] = 5
plt.rcParams['ytick.minor.size'] = 5
plt.rcParams['xtick.major.width'] = 2
plt.rcParams['ytick.major.width'] = 2
plt.rcParams['xtick.minor.width'] = 1
plt.rcParams['ytick.minor.width'] = 1
plt.rcParams['xtick.top'] = True
plt.rcParams['ytick.right'] = True
plt.rcParams['xtick.minor.visible'] = True
plt.rcParams['ytick.minor.visible'] = True
```

```python
L = 10
h = 0.0
temp = np.linspace(2.269, 5, 10)
```

```python
new_distances = []
for i in range(len(temp)):
    new_distances.append(np.loadtxt(f"Hamming_Distances/new_distances_T_{temp[i]}.txt"))
```

```python
new_dist_5 = np.loadtxt(f"Hamming_Distances/new_distances_T_5.0.txt")
new_dist_Tc = np.loadtxt(f"Hamming_Distances/new_distances_T_2.269.txt")
```

```python
plt.hist(new_dist_5, bins =  20, density = True)
plt.title(f"T = {temp[-1]}")
plt.show()

plt.hist(new_distances[3], bins =  17, density = True)
plt.title(f"T = {temp[-4]}")
plt.show()

plt.hist(new_distances[5], bins =  18, density = True)
plt.title(f"T = {temp[-6]}")
plt.show()

plt.hist(new_distances[7], bins =  16, density = True)
plt.title(f"T = {temp[-8]}")
plt.show()

plt.hist(new_dist_Tc, bins =  16, density = True)
plt.title(f"T = {temp[-10]}")
plt.show()
```

```python
plt.figure(figsize=(10, 7))

plt.subplot(2, 2, 1)
plt.hist(new_dist_5, bins=20, density=True)
plt.title(f"T = {temp[-1]:.2f}")
plt.xlabel(r"$d$", fontsize=15)

plt.subplot(2, 2, 2)
plt.hist(new_distances[3], bins=17, density=True)
plt.title(f"T = {temp[-4]:.2f}")
plt.xlabel(r"$d$", fontsize=15)

plt.subplot(2, 2, 3)
plt.hist(new_distances[7], bins=16, density=True)
plt.title(f"T = {temp[-8]:.2f}")
plt.xlabel(r"$d$", fontsize=15)

plt.subplot(2, 2, 4)
plt.hist(new_dist_Tc, bins=16, density=True)
plt.title(f"T = {temp[-10]:.2f}")
plt.xlabel(r"$d$", fontsize=15)

plt.tight_layout()
plt.show()
```

```python
indices = [0, 3, 7, 9]
bins = [20, 17, 16, 16]
titles = [r"$T = {0:.2f}$".format(temp[-1]), 
          r"$T = {0:.2f}$".format(temp[-4]), 
          r"$T = {0:.2f}$".format(temp[-8]), 
          r"$T = {0:.2f}$".format(temp[-10])]

plt.figure(figsize=(10, 8))
for i, index in enumerate(indices):
    plt.subplot(2, 2, i+1)
    plt.hist(new_distances[index], bins=bins[i], density=True)
    plt.title(titles[i], fontsize = 14)
    if i > 1:
        plt.xlabel(r"$d$", fontsize = 15)
```
